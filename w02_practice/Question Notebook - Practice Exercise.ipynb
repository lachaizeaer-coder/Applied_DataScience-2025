# Binomial Distribution

### Q1. A LED bulb manufacturing company regularly conducts quality checks at specified periods on the products it manufactures.  Historically, the failure rate for LED light bulbs that the company manufactures is 5%. Suppose a random sample of 10 LED light bulbs is selected. Find the probability distribution for the defective bulbs and answer the following questions.

```python
p = 0.05
n = 10
k = np.arange(0, 11)
```

```python
binomial = stats.binom.pmf(k, n, p)
binomial
```

```python
#plot the distribution
plt.bar(k, binomial)
plt.xlabel('Number of defective bulbs')
plt.ylabel('Probability')
plt.title('Binomial Distribution')
plt.show()
```

**a) What is the probability that none of the LED bulbs are defective?**

```python
print('The probability that none of the LED bulbs are defective is ~ ' + str( round( binomial[0], 2 )) + ' (or ' + str( round( binomial[0], 2 ) * 100) + "%)" )
```

**b) What is the probability that exactly one LED bulb is defective?**

```python
print('The probability that one of the LED bulbs are defective is ~ ' + str( round( binomial[1], 2 )) + ' (or ' + str( round( binomial[1], 2 ) * 100) + "%)" )
```

**c) What is the probability that two or fewer LED bulbs are defective?**

Hint: We need to calculate the cumulative probability of two or fewer LED bulbs being defective.

```python
cumulative_binomial = stats.binom.cdf(2, n, p)
print('The probability that two or more of the LED bulbs are defective is ~ ' + str( round( cumulative_binomial, 4 )) + ' (or ' + str( round( cumulative_binomial, 2 ) * 100) + "%)" )
```

**d) What is the probability that three or more LED bulbs are defective?**

Hint: We need to subtract the cumulative probability of up to 2 defective LED bulbs from 1.

```python
cumulative_binomial_complement = 1 - stats.binom.cdf(2, n, p)
print('The probability that three or more of the LED bulbs are defective is ~ ' + str( round( cumulative_binomial_complement, 4 )) + ' (or ' + str( round( cumulative_binomial_complement, 3 ) * 100) + "%)" )
```

### Q3.	Over a long period in a large multinomial corporation, 10% of all sales trainees rated as outstanding, 75% rated as excellent, 10% rated as satisfactory, and 5% are considered unsatisfactory. Find the following probabilities for a sample of 10 trainees selected at random.

**a) Two are rated as outstanding**

```python
p = 0.1
n = 10
k = 2

binomial = stats.binom.pmf(k, n, p)
print('The probability that two of the ten trainees are rated as outstanding is ~ ' + str( round( binomial, 2 )) + ' (or ' + str( int(round( binomial, 2 ) * 100)) + "%)" )
```

**b) Two or more are rated as outstanding**

```python
binomial_cumulative = stats.binom.cdf(1, n, p)
print('The probability that two or more of the ten trainees are rated as outstanding is ~ ' + str( round( 1 - binomial_cumulative, 2 )) + ' (or ' + str( int(round( 1-binomial_cumulative, 2 ) * 100)) + "%)")
```

**c) Eight of the ten are rated either outstanding or excellent**

```python
p = 0.85
n = 10
k = 8

binomial = stats.binom.pmf(k, n, p)
print('The probability that eight of the ten trainees are rated either outstanding or excellent is ~ ' + str( round( binomial, 2 )) + ' (or ' + str( int(round( binomial, 2 ) * 100)) + "%)" )
```

**d) None of the trainees are rated as unsatisfactory**

```python
p = 0.05
n = 10
k = 0

binomial = stats.binom.pmf(k, n, p)
print('The probability that none of the ten trainees are rated as unsatisfactory is ~ ' + str( round( binomial, 4 )) + ' (or ' + str( int(round( binomial, 2 ) * 100)) + "%)" )
```

**a) Find the probability that a randomly selected student requires at most 2.5 hours to complete the assignment.**

**CDF:** of a random variable (X) is the probability that X will take the value less than or equal to x. It can be represented mathematically as below:

>$F_X(x) = P(X\leq x)$

In our case, the random variable (X) is the number of hours.

$ P(X\leq 2.5)$

```python
probability = uniform.cdf(2.5, loc=0, scale=3)
print('The probability that a randomly selected student requires at most 2.5 hours to complete the assignment is ~ ' + str( round( probability, 2 )) + ' (or ' + str( int(round( probability, 3 ) * 100)) + "%)" )
```

**b) Find the probability that a randomly selected student requires at least 3 hours to complete the quiz.**

$ P(X>=3)$

```python
probability = 1- uniform.cdf(2.5, loc=0, scale=3)
print('The probability that a randomly selected student requires at least 3 hours to complete the assignment is ~ ' + str( round( probability, 2 )) + ' (or ' + str( int(round( probability, 3 ) * 100)) + "%)" )
```

**c) Find the probability that a randomly selected student requires 1.5 to 3.5 hours to complete the quiz.**

$ P(1.5<= X <=3.5)$

```python
probability = uniform.cdf(3.5, loc=0, scale=3) - uniform.cdf(1.5, loc=0, scale=3)
print('The probability that a randomly selected student requires 1.5 to 3.5 hours to complete the assignment is ~ ' + str( round( probability, 2 )) + ' (or ' + str( int(round( probability, 3 ) * 100)) + "%)")
```
